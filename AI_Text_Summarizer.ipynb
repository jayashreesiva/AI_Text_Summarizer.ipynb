```python
import nltk
import re
from collections import Counter
```

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
text = """
Artificial Intelligence is a branch of computer science that focuses on
creating machines and computer systems that can perform tasks that normally
require human intelligence. These tasks include learning, reasoning,
problem-solving, understanding natural language, and recognizing patterns.

Artificial Intelligence is widely used in many fields. In healthcare, AI can
help doctors analyze medical images and support disease detection. In education,
AI can provide personalized learning experiences for students. In transportation,
AI is used in navigation systems, traffic prediction, and autonomous vehicles.

Machine Learning is an important part of Artificial Intelligence. It allows
computers to learn from data and improve their performance without being
explicitly programmed for every task.
"""

print("Original Text:")
print(text)
```

```python
# Convert text to lowercase
clean_text = text.lower()

# Remove extra spaces and new lines
clean_text = re.sub(r'\s+', ' ', clean_text).strip()

print("Cleaned Text:")
print(clean_text)
```

```python
# Split the text into sentences
sentences = nltk.sent_tokenize(text)

print("Total Number of Sentences:", len(sentences))
print("\nSentences:")
for i, sentence in enumerate(sentences, 1):
    print(f"{i}. {sentence}")
```

```python
from nltk.corpus import stopwords

# Get English stopwords
stop_words = set(stopwords.words('english'))

# Find all words
words = re.findall(r'\b[a-zA-Z]+\b', clean_text)

# Remove common stopwords
filtered_words = [
    word for word in words
    if word not in stop_words
]

# Count word frequency
word_frequency = Counter(filtered_words)

print("Most Important/Frequent Words:")
print("-------------------------------")

for word, count in word_frequency.most_common(10):
    print(word, ":", count)
```

```python
# Calculate a score for each sentence

sentence_scores = {}

for sentence in sentences:
    words_in_sentence = re.findall(r'\b[a-zA-Z]+\b', sentence.lower())

    score = 0

    for word in words_in_sentence:
        if word in word_frequency:
            score += word_frequency[word]

    sentence_scores[sentence] = score

# Display sentence scores
print("Sentence Scores:")
print("-------------------------")

for sentence, score in sentence_scores.items():
    print("Score:", score)
    print(sentence)
    print()
```

```python
# Number of sentences we want in the summary
summary_length = 3

# Sort sentences based on their scores
ranked_sentences = sorted(
    sentence_scores,
    key=sentence_scores.get,
    reverse=True
)

# Select the top sentences
selected_sentences = ranked_sentences[:summary_length]

# Keep the original order of the selected sentences
summary_sentences = [
    sentence for sentence in sentences
    if sentence in selected_sentences
]

# Create the final summary
summary = " ".join(summary_sentences)

print("📄 AI TEXT SUMMARY")
print("----------------------------")
print(summary)
```

```python
# Count words in original text
original_words = re.findall(r'\b[a-zA-Z]+\b', text)
original_word_count = len(original_words)

# Count words in summary
summary_words = re.findall(r'\b[a-zA-Z]+\b', summary)
summary_word_count = len(summary_words)

# Calculate compression percentage
if original_word_count > 0:
    compression_percentage = (
        (original_word_count - summary_word_count)
        / original_word_count
    ) * 100
else:
    compression_percentage = 0

print("📊 TEXT SUMMARY STATISTICS")
print("----------------------------")
print("Original Word Count :", original_word_count)
print("Summary Word Count  :", summary_word_count)
print("Compression         :", round(compression_percentage, 2), "%")

print("\n📄 ORIGINAL TEXT")
print("----------------------------")
print(text)

print("\n📝 SUMMARY")
print("----------------------------")
print(summary)
```

```python
# Use the extracted PDF text
text = pdf_text

# Clean the text
clean_text = text.lower()
clean_text = re.sub(r'\s+', ' ', clean_text).strip()

# Split into sentences
sentences = nltk.sent_tokenize(text)

# Extract words
words = re.findall(r'\b[a-zA-Z]+\b', clean_text)

# Remove stopwords
stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in words
    if word not in stop_words
]

# Count word frequency
word_frequency = Counter(filtered_words)

print("✅ PDF text processed successfully!")
print("Number of sentences:", len(sentences))
print("Number of important words:", len(word_frequency))
```

```python
# Calculate importance score for each sentence
sentence_scores = {}

for sentence in sentences:
    sentence_words = re.findall(
        r'\b[a-zA-Z]+\b',
        sentence.lower()
    )

    score = 0

    for word in sentence_words:
        if word in word_frequency:
            score += word_frequency[word]

    sentence_scores[sentence] = score


# Select important sentences
summary_length = min(5, len(sentences))

ranked_sentences = sorted(
    sentence_scores,
    key=sentence_scores.get,
    reverse=True
)

selected_sentences = ranked_sentences[:summary_length]


# Keep the original order of sentences
summary_sentences = [
    sentence for sentence in sentences
    if sentence in selected_sentences
]

summary = " ".join(summary_sentences)


print("=" * 60)
print("        📄 AI TEXT SUMMARIZER")
print("=" * 60)

print("\n📝 GENERATED SUMMARY")
print("-" * 60)
print(summary)
```
